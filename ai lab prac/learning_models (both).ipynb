```python
# LINEAR RERESSION 

# ================== SAMPLE CSV DATASET ==================
# Save this as data.csv if needed:
#
# hours,marks
# 1,40
# 2,50
# 3,60
# 4,70
# 5,80
# =======================================================


# Step 1: Import libraries
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score


# Step 2: Load dataset
data = pd.read_csv("data.csv")   # <-- change file name if needed

# Step 3: Define x (input) and y (output)
# Change column names according to your dataset
x = data[['hours']]   # independent variable (must be 2D)
y = data['marks']     # dependent variable


# Step 4: Split dataset
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
) # test_size=0.2 means 20% test data, 80% training


# Step 5: Create model
model = LinearRegression()


# Step 6: Train model
model.fit(x_train, y_train)


# Step 7: Predict
y_pred = model.predict(x_test)


# Step 8: Evaluation

# Mean Squared Error
mse = mean_squared_error(y_test, y_pred)

# Residuals (error)
residuals = y_pred - y_test

# R2 Score (testing accuracy)
r2 = r2_score(y_test, y_pred)
accuracy = r2 * 100


# Step 9: Print results
print("\n=================== RESULTS ===================")
print("Predictions:", y_pred)
print("Actual:", list(y_test))
print("Residuals:", list(residuals))

print("\nMSE:", mse)
print("R2 Score:", r2)
print("Testing Accuracy (%):", accuracy)


# Step 11: Model parameters (important for viva)
print("\nCoefficient (Slope):", model.coef_)
print("Intercept:", model.intercept_)

```

```python
# LINEAR RERESSION with K-FOLDS CV

# ================== SAMPLE CSV DATASET ==================
# Save this as data.csv if needed:
#
# hours,marks
# 1,40
# 2,50
# 3,60
# 4,70
# 5,80
# 6,90
# 7,100
# =======================================================


# Step 1: Import libraries
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import KFold, cross_val_score


# Step 2: Load dataset
data = pd.read_csv("data.csv")

# Step 3: Define x and y
x = data[['hours']]   # input (2D)
y = data['marks']     # output


# Step 4: Create model
model = LinearRegression()


# Step 5: Apply K-Fold Cross Validation
kf = KFold(n_splits=5, shuffle=True, random_state=42)

# R2 scores
r2_scores = cross_val_score(model, x, y, cv=kf, scoring='r2')

# MSE scores
mse_scores = cross_val_score(model, x, y, cv=kf, scoring='neg_mean_squared_error')
mse_scores = -mse_scores  # convert to positive


# Step 6: Print results
print("\n=================== K-FOLD RESULTS ===================")
print("R2 Scores for each fold:", r2_scores)
print("Average R2:", r2_scores.mean())
print("Average Accuracy (%):", r2_scores.mean() * 100)

print("\nMSE for each fold:", mse_scores)
print("Average MSE:", mse_scores.mean())
```

```python
# DECISION TREE
# ================== IMPORT LIBRARIES ==================
import numpy as np
import pandas as pd

from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score


# ================== LOAD DATASET ==================
# Example CSV dataset format:
#
# age, income, buys
# 25, 50000, 0
# 30, 60000, 0
# 35, 80000, 1
# 40, 90000, 1
#

data = pd.read_csv("data.csv")

print("First 5 rows:\n", data.head())
print("\nColumns:", data.columns)


# ================== DEFINE X AND Y ==================
# Replace column names based on your dataset

x = data[['age', 'income']]   # features (input)
y = data['buys']              # target (output)


# ================== TRAIN TEST SPLIT ==================
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)


# ================== CREATE MODEL ==================
DT = DecisionTreeClassifier()


# ================== TRAIN MODEL ==================
ModelDT = DT.fit(x_train, y_train)


# ================== PREDICTION ==================
PredictionDT = ModelDT.predict(x_test)

print("\nPredictions:", PredictionDT)


# ================== TRAINING ACCURACY ==================
print('\n==================== DT TRAINING ACCURACY ====================')

training_acc = DT.score(x_train, y_train)
TrainingAccDT = training_acc * 100

print(f"Training Accuracy: {TrainingAccDT:.2f}%")


# ================== TESTING ACCURACY ==================
print('\n==================== DT TESTING ACCURACY ====================')

testing_acc = accuracy_score(y_test, PredictionDT)
TestingAccDT = testing_acc * 100

print(f"Testing Accuracy: {TestingAccDT:.2f}%")
```

```python
# K-MODEL CLUSTERING

# ================== IMPORT LIBRARIES ==================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler


# ================== LOAD DATASET ==================
# Example CSV format:
#
# feature1, feature2
# 1.2, 3.4
# 2.1, 4.5
# 3.3, 1.8
#

data = pd.read_csv("data.csv")

print("First 5 rows:\n", data.head())


# ================== FEATURE SELECTION ==================
# Select only numerical features
x = data[['feature1', 'feature2']]


# ================== FEATURE SCALING (IMPORTANT) ==================
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)


# ================== ELBOW METHOD (FIND BEST K) ==================
wcss = []

for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(x_scaled)
    wcss.append(kmeans.inertia_)   # WCSS

# Plot Elbow graph
plt.plot(range(1, 11), wcss)
plt.title("Elbow Method")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS")
plt.show()


# ================== APPLY K-MEANS ==================
# Suppose best K = 3 (choose from elbow graph)
kmeans = KMeans(n_clusters=3, random_state=42)
y_kmeans = kmeans.fit_predict(x_scaled)


# ================== ADD CLUSTER LABELS ==================
data['Cluster'] = y_kmeans

print("\nClustered Data:\n", data.head())


# ================== VISUALIZATION ==================
plt.scatter(x_scaled[:, 0], x_scaled[:, 1], c=y_kmeans, cmap='viridis')

# Cluster centers
centers = kmeans.cluster_centers_
plt.scatter(centers[:, 0], centers[:, 1], s=200, c='red')

plt.title("K-Means Clustering Result")
plt.show()
```

```python
# LOGISTIC REGRESSION

# 1. Import libraries
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report


# 2. Load CSV file
df = pd.read_csv("data.csv")


# 3. View dataset (important in exams)
print(df.head())
print(df.columns)


# 4. Split into input (X) and output (y)
# Change column names according to your dataset

x = df[["feature1", "feature2"]]   # input features
y = df["label"]                    # output (0 or 1)


# 5. Train-test split
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)


# 6. Create model
model = LogisticRegression()


# 7. Train model
model.fit(x_train, y_train)


# 8. Make predictions
y_pred = model.predict(x_test)


# 9. Evaluate model

accuracy = accuracy_score(y_test, y_pred)

print("\n================ ACCURACY ================")
print("Accuracy:", accuracy * 100)

print("\n================ CONFUSION MATRIX ================")
print(confusion_matrix(y_test, y_pred))

print("\n================ CLASSIFICATION REPORT ================")
print(classification_report(y_test, y_pred))


# 10. Predict new unseen data
new_data = [[0.3, 0.8]]   # example feature values
prediction = model.predict(new_data)

print("\nPrediction (0=Not Spam, 1=Spam):", prediction)
```

```python
# LOGISTIC REGRESSION with K-FOLD

import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import KFold, cross_val_score

# Load dataset
df = pd.read_csv("data.csv")

# Features and target
x = df[["feature1", "feature2"]]
y = df["label"]

# Model
model = LogisticRegression()

# K-Fold setup
kf = KFold(n_splits=5, shuffle=True, random_state=42)

# Cross-validation (accuracy scoring)
scores = cross_val_score(model, x, y, cv=kf, scoring='accuracy')

print("Accuracy per fold:", scores)
print("Average Accuracy:", scores.mean() * 100)
```
