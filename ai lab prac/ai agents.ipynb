```python
# SIMPLE REFLEX AGENT 1D GRIDs

class Environment:
    def __init__(self):
        self.grid = ['clean','dirty','clean',
                    'clean','clean','dirty',
                    'dirty','clean','dirty']       

    def get_percept(self,position):
        return self.grid[position]

    def cleanRoom(self,position):
        self.grid[position] = 'clean'

    def display(self,agent_pos):
        #displays grid with agent position
        display_grid = self.grid[:]
        display_grid[agent_pos] = "X"

        for i in range(0,9,3):
            print(' | '.join(display_grid[i:i+3]))
        print() # extra line

#envir = Environment().display(3)

class SimpleReflexAgent:
    def __init__(self):
        self.position = 0
    
    def act(self,percept,grid):
        if percept == 'dirty':
            grid[self.position] = 'clean'
            return 'room cleaned'
        else:
            return 'room already clean'

    def move(self):
        if self.position <8:
            self.position += 1
        return self.position


def run_agent(agent,environment,steps):
    for i in range(steps):
        percept = environment.get_percept(agent.position)
        action = agent.act(percept, environment.grid)

        print(f"Step {i+1} : Position {agent.position} -> Percept {percept}, Action - {action}")

        environment.display(agent.position)

        if percept == 'dirty':
            environment.cleanRoom(agent.position)

        agent.move()

agent = SimpleReflexAgent()
environment = Environment()

run_agent(agent,environment,9)
            


```

```python
# BEST FIRST SEARCH = UNIFORM COST SEARCH
from queue import PriorityQueue 
graph = {
    'S': [('A', 3), ('B', 6), ('C', 5)],
    'A': [('D', 9), ('E', 8)],
    'B': [('F', 12), ('G', 14)],
    'C': [('H', 7)],
    'H': [('I', 5), ('J', 6)],
    'I': [('K', 1), ('L', 10), ('M', 2)],
    'D': [], 'E': [],
    'F': [], 'G': [],
    'J': [], 'K': [],
    'L': [], 'M': []
}
def UCS(graph,start,goal):
    visited = set()
    pq = PriorityQueue()
    pq.put((0,start,[start])) #[total cost,start node,path]

    while not pq.empty():
        total_cost,node,path = pq.get()

        if node == goal:
            print("Goal Node Found! path: ",path)
            return path,total_cost

        visited.add(node)

        for neighbour,weight in graph[node]:
            if neighbour not in visited:
                pq.put((total_cost+weight,neighbour,path+[neighbour]))

    print("goal not found")
    return None,float('inf')

print("UNIFORM COST SEARCH")
UCS(graph,'S','M')


heuristic = {
    'S': 4,
    'A': 4,
    'B': 4,
    'C': 3,
    'D': 4,
    'E': 2,
    'F': 4,
    'G': 3,
    'H': 3,
    'I': 2,
    'J': 3,
    'K': 1,
    'L': 2,
    'M': 0
}


# GREEDY BEST FIRST SEARCH
def GBFS(graph,start,goal,heuristic_func):
    visited = set()
    pq = PriorityQueue()
    pq.put((heuristic_func[start],start,[start])) #[node_neuristic,start node,path]
    

    while not pq.empty():
        h,node,path = pq.get()

        if node == goal:
            print("Goal Node Found! path: ",path)
            return path,"mr. GREEDY"

        visited.add(node)

        for neighbour,_ in graph[node]:
            if neighbour not in visited:
                pq.put((heuristic[neighbour],neighbour,path+[neighbour]))

    print("goal not found")
    return None,float('inf')

print("GREEDY BEST FIRST SEARCH")
GBFS(graph,'S','M',heuristic)
```

```python
import random

# initial population
def generate_population(size):
    return [''.join(random.choice('01') for _ in range(5)) for _ in range(size)]

# decode
def decode(population):
    return int(population,2)

# fitness
def fitness(population):
    x = decode(population)
    return x*x + x

# selection
def selection(population):
    population = sorted(population, key=lambda x: fitness(x), reverse=True)
    return population[:2]

# crossover
def crossover(p1,p2):
    point = random.randint(1,4)
    c1 = p1[:point] + p2[point:]
    c2 = p2[:point] + p1[point:]
    return c1,c2

# mutation
def mutation(population):
    i = random.randint(0,4)
    population = list(population)
    population[i] = '0' if population[i]=='1' else '1'
    return ''.join(population)

# GA
def GA():
    population = generate_population(6)

    for gen in range(10):
        print(f"\nGeneration {gen}: {population}")

        p1,p2 = selection(population)

        new_population = [p1,p2]

        while len(new_population) < 6:
            c1,c2 = crossover(p1,p2)
            c1 = mutation(c1)
            c2 = mutation(c2)
            new_population.extend([c1,c2])

        population = new_population[:6]   # ✅ FIX

    # final best (after evolution)
    best = max(population, key=lambda x: fitness(x))
    x = decode(best)

    print("\nBest Binary:", best)
    print("Best x:", x)
    print("Maximum f(x)=2x+1:", 2*x + 1)

GA()
```

```python
# past paper q2

from queue import PriorityQueue

# Define the grid
grid = [
    [1, 1, 2, '#', 3],
    [1, '#', 2, '#', '#'],
    [1, 2, 1, 2, '#'],
    ['#', '#', 1, '#', '#'],
    [1, 1, 1, 2, 1]
]

rows, cols = len(grid), len(grid[0])
start = (0, 0)
goal = (rows-1, cols-1)

# Manhattan distance heuristic
def manhattan(cell):
    r, c = cell
    gr, gc = goal
    return abs(r - gr) + abs(c - gc)

# A* function similar to your graph A* code
def A_star_grid(grid, start, goal):
    pq = PriorityQueue()
    pq.put((manhattan(start), 0, start, [start]))  # (f, g, node, path)
    cost_so_far = {start: 0}

    # Moves: up, down, left, right
    moves = [(-1,0),(1,0),(0,-1),(0,1)]

    while not pq.empty():
        f, g, node, path = pq.get()
        r, c = node

        if node == goal:
            print("Goal found!")
            print("Path:", path)
            print("Total cost:", g)
            return path, g

        for dr, dc in moves:
            nr, nc = r + dr, c + dc
            if 0 <= nr < rows and 0 <= nc < cols:
                if grid[nr][nc] != '#':
                    g_new = g + grid[nr][nc]  # cost to move to neighbor
                    h_new = g_new + manhattan((nr,nc))
                    if (nr,nc) not in cost_so_far or g_new < cost_so_far[(nr,nc)]:
                        cost_so_far[(nr,nc)] = g_new
                        pq.put((h_new, g_new, (nr,nc), path + [(nr,nc)]))

    print("Goal not reachable")
    return None, float('inf')

# Run A* on the grid
A_star_grid(grid, start, goal)
```

```python
# Iterative Deepening DFS
graph = {
    'A': ['B','C'],
    'B': ['D','E'],
    'C': ['F'],
    'D': ['F'],
    'E': [],
    'F': []
}
# Depth Limited DFS
def DLS(node, goal, limit, path):
    print(node, end=" -> ")

    if node == goal:
        print("GOAL FOUND!")
        print("PATH =", path)
        return True

    if limit <= 0:
        return False

    for neighbour in graph[node]:
        if neighbour not in path:   # avoid cycles
            if DLS(neighbour, goal, limit-1, path + [neighbour]):
                return True
    return False

# IDDFS
def IDDFS(start, goal, max_depth):
    for depth in range(max_depth + 1):
        print(f"\nDepth Limit = {depth}")
        if DLS(start, goal, depth, [start]):
            return
    print("Goal not found")
# Run
IDDFS('A', 'F', 5)
```
